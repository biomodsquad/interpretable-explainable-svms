# Training-only selection of SVM-set versus unified prediction

This experiment uses a nested architecture-validation split drawn exclusively from the training data. It compares the current MiSTIC SVM-set prediction with a single kernel-matched sklearn SVM trained on the knee-ranked unified features. The winning architecture is selected without observing the fixed blind set, both candidates are rebuilt on all available training data, and the selected prediction is evaluated once on the same blind set. Classification uses the current 5-member RBF/weight-0.90 configuration; harder regression uses the current 20-member linear/weight-0.75 configuration.

```python
from pathlib import Path
import sys
repo_root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'mistic' / 'svmSet.py').exists())
example_dir = repo_root / 'mistic' / 'examples'
if str(example_dir) not in sys.path: sys.path.insert(0, str(example_dir))
from synthetic100_nested_unified_selection import run_experiment
results = run_experiment()
results
```

```python
classification_summary = results.query("task == 'classification'").groupby('method')[['roc_auc', 'f1', 'balanced_accuracy', 'accuracy']].agg(['mean', 'std'])
classification_summary.round(4)
```

```python
regression_summary = results.query("task == 'regression'").groupby('method')[['r2', 'rmse', 'mae', 'pearson_r2']].agg(['mean', 'std'])
regression_summary.round(4)
```

```python
selection = (results.query("method == 'training_selected'")
             .groupby(['task', 'selected_architecture']).size()
             .rename('count').to_frame())
selection
```

```python
paired = []
for task, metric in [('classification', 'roc_auc'), ('regression', 'r2')]:
    wide = results.query('task == @task').pivot(index='seed', columns='method', values=metric)
    for method in ['unified', 'training_selected']:
        delta = wide[method] - wide['svm_set']
        paired.append({'task': task, 'method_vs_set': method, 'mean_difference': delta.mean(),
                       'wins': (delta > 0).sum(), 'ties': (delta == 0).sum(), 'losses': (delta < 0).sum()})
import pandas as pd
pd.DataFrame(paired)
```
